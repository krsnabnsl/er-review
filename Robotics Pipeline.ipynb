```python
from PIL import Image, ImageDraw, ImageFont, ImageFilter
import os, math

# ---------- Global config (same as before) ----------
VALUE_MIN, VALUE_MAX = 0, 100
ANGLE_MIN, ANGLE_MAX = -210, 30  # where min/max values land on the dial (graphics degrees)

W, H = 1000, 800
SCALE = 4
w, h = W*SCALE, H*SCALE

font_paths = [
    "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
    "/usr/share/fonts/truetype/dejavu/DejaVuSansCondensed.ttf",
]
FONT = next((p for p in font_paths if os.path.exists(p)), None)

def deg2rad(d): return d*math.pi/180.0

def value_to_angle(v, value_min=VALUE_MIN, value_max=VALUE_MAX, angle_min=ANGLE_MIN, angle_max=ANGLE_MAX):
    v = max(min(v, value_max), value_min)
    k = (v - value_min) / (value_max - value_min)
    return angle_min + k*(angle_max - angle_min)

def convert_robot_angle(a_deg, system="graphics"):
    """
    Convert various angle conventions to Pillow's (+x axis to the right, CCW positive).
    system:
      - 'graphics'            : already +x, CCW
      - 'robot_ccw_from_top'  : 0° at 12 o'clock (+y), increasing CCW
      - 'robot_cw_from_top'   : 0° at 12 o'clock, increasing CW
    """
    if system == "graphics":
        return a_deg
    elif system == "robot_ccw_from_top":
        return a_deg - 90
    elif system == "robot_cw_from_top":
        return -a_deg - 90
    else:
        return a_deg

def polar(cx, cy, r, ang_deg):
    a = deg2rad(ang_deg)
    return cx + r*math.cos(a), cy + r*math.sin(a)

def draw_gauge_priority_value(
    filename,
    label,
    value,
    angle=None,                       # optional robot-provided angle
    unit=None,
    value_min=VALUE_MIN,
    value_max=VALUE_MAX,
    angle_min=ANGLE_MIN,
    angle_max=ANGLE_MAX,
    robot_angle_system="graphics",    # adjust if your er_angle isn't graphics-based
    show_angle_marker=True,           # draw a thin marker for the provided angle
    show_mismatch_text=True           # display small Δdeg if angle is provided
):
    # ---------- Layout ----------
    img = Image.new("RGBA", (w, h), (245, 247, 250, 255))
    d = ImageDraw.Draw(img)

    cx, cy = w//2, int(h*0.58)
    R_outer = int(min(w, h)*0.36)
    R_inner = R_outer - int(30*SCALE)

    # soft drop shadow
    shadow = Image.new("RGBA", img.size, (0,0,0,0))
    sd = ImageDraw.Draw(shadow)
    sd.ellipse([cx-R_outer-10*SCALE, cy-R_outer-10*SCALE,
                cx+R_outer+10*SCALE, cy+R_outer+10*SCALE], fill=(0,0,0,100))
    shadow = shadow.filter(ImageFilter.GaussianBlur(12*SCALE))
    img = Image.alpha_composite(img, shadow); d = ImageDraw.Draw(img)

    # bezel + dial
    d.ellipse([cx-R_outer, cy-R_outer, cx+R_outer, cy+R_outer],
              fill=(230,232,236,255), outline=(60,60,60,255), width=6*SCALE)
    d.ellipse([cx-R_inner, cy-R_inner, cx+R_inner, cy+R_inner],
              fill=(250,250,252,255), outline=(180,180,180,255), width=4*SCALE)

    # glare
    glare = Image.new("RGBA", img.size, (0,0,0,0))
    gd = ImageDraw.Draw(glare)
    gd.pieslice([cx-R_inner, cy-R_inner, cx+R_inner, cy+R_inner],
                start=200, end=340, fill=(255,255,255,40))
    img = Image.alpha_composite(img, glare); d = ImageDraw.Draw(img)

    # colored arcs
    arc_r = R_inner - 18*SCALE
    band_w = 14*SCALE
    def arc(start, end, color):
        d.arc([cx-arc_r, cy-arc_r, cx+arc_r, cy+arc_r], start=start, end=end,
              fill=color, width=band_w)
    # tune these ranges if needed
    arc(angle_min, -90, (46, 204, 113, 255))  # green
    arc(-90, -30, (241, 196, 15, 255))        # yellow
    arc(-30, angle_max, (231, 76, 60, 255))   # red

    # ticks + numbers
    major_every, minor_every = 10, 2
    tick_outer = R_inner - 8*SCALE
    major_len, minor_len = 28*SCALE, 15*SCALE

    if FONT:
        font_major = ImageFont.truetype(FONT, 28*SCALE)
        font_label = ImageFont.truetype(FONT, 34*SCALE)
        font_value = ImageFont.truetype(FONT, 40*SCALE)
        font_small = ImageFont.truetype(FONT, 22*SCALE)
    else:
        font_major = font_label = font_value = font_small = None

    for v in range(value_min, value_max+1, minor_every):
        a = value_to_angle(v, value_min, value_max, angle_min, angle_max)
        x1, y1 = polar(cx, cy, tick_outer, a)
        L = major_len if (v % major_every == 0) else minor_len
        x2, y2 = polar(cx, cy, tick_outer - L, a)
        d.line([(x1,y1),(x2,y2)], fill=(30,30,30,255), width=3*SCALE)

        if v % major_every == 0:
            xn, yn = polar(cx, cy, tick_outer - L - 28*SCALE, a)
            text = str(v)
            if font_major:
                bbox = d.textbbox((0,0), text, font=font_major)
                tw, th = bbox[2]-bbox[0], bbox[3]-bbox[1]
            else:
                tw, th = 20*SCALE, 20*SCALE
            d.text((xn - tw/2, yn - th/2), text, fill=(20,20,20,255),
                   font=font_major)

    # --- PRIORITY: needle from VALUE ---
    a_value = value_to_angle(value, value_min, value_max, angle_min, angle_max)

    # optional: show the raw 'angle' as a thin ring marker
    mismatch_deg = None
    if angle is not None and show_angle_marker:
        a_robot = convert_robot_angle(angle, robot_angle_system)
        # marker
        marker_r_outer = R_inner - 10*SCALE
        marker_r_inner = marker_r_outer - 6*SCALE
        x_out, y_out = polar(cx, cy, marker_r_outer, a_robot)
        x_in,  y_in  = polar(cx, cy, marker_r_inner, a_robot)
        d.line([(x_in, y_in), (x_out, y_out)], fill=(100,100,100,255), width=6*SCALE)
        # compute mismatch (for debugging / display)
        mismatch_deg = (a_robot - a_value + 540) % 360 - 180  # wrap to [-180,180]

    # needle shadow
    needle_len = R_inner - 38*SCALE
    x_tip, y_tip = polar(cx, cy, needle_len, a_value)
    d.line([(cx+3*SCALE, cy+3*SCALE), (x_tip+3*SCALE, y_tip+3*SCALE)],
           fill=(0,0,0,120), width=10*SCALE)

    # needle (from value)
    d.line([(cx, cy), (x_tip, y_tip)], fill=(214,48,49,255),
           width=10*SCALE, joint="curve")

    # hub
    cap_r1, cap_r2 = 24*SCALE, 12*SCALE
    d.ellipse([cx-cap_r1, cy-cap_r1, cx+cap_r1, cy+cap_r1], fill=(40,40,40,255))
    d.ellipse([cx-cap_r2, cy-cap_r2, cx+cap_r2, cy+cap_r2], fill=(230,230,230,255))

    # labels
    value_text = f"{value:.2f}" + (f" {unit}" if unit else "")
    d.text((80*SCALE, 70*SCALE), f"{label}", fill=(20,20,20,255), font=font_label)
    d.text((80*SCALE, 120*SCALE), value_text, fill=(60,60,60,255), font=font_value)

    if mismatch_deg is not None and show_mismatch_text:
        d.text((80*SCALE, 170*SCALE), f"Δ angle: {mismatch_deg:+.1f}°",
               fill=(120,120,120,255), font=font_small)

    # frame + downscale
    d.rectangle((40*SCALE, 40*SCALE, w-40*SCALE, h-40*SCALE),
                outline=(90,90,90,255), width=6*SCALE)

    out = img.resize((W, H), Image.LANCZOS).convert("RGB")
    out.save(filename, format="PNG")
```

```python
BASE_DIR = "/Volumes/roboticscatalog/default/roboticsvolume/photos"
os.makedirs(BASE_DIR, exist_ok=True)

samples = [
    # (photo, er_name, er_value, er_angle, site, date, mission_id)
    ("gauge1.png", "manometer_0", 18.95, 108.63),
    ("gauge2.png", "manometer_1", 42.10,  12.00),
    ("gauge3.png", "manometer_2",  7.45,  85.20),
    ("gauge4.png", "manometer_3", 63.30,  33.75),
    ("gauge5.png", "manometer_4", 29.80, 274.10),
    ("gauge6.png", "manometer_5", 55.60,  47.30),
    ("gauge7.png", "manometer_6", 11.25, 190.50),
    ("gauge8.png", "manometer_7", 38.70,  62.15),
]

for fname, label, val, ang in samples:
    draw_gauge_priority_value(
        filename=os.path.join(BASE_DIR, fname),
        label=label,
        value=val,
        angle=ang,                       # supplied, but used only as marker
        robot_angle_system="robot_ccw_from_top"  # <-- set this correctly for your er_angle
    )
```
