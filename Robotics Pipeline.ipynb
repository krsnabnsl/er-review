```python
from PIL import Image, ImageDraw
import os, math

base_dir = "/Volumes/roboticscatalog/default/roboticsvolume/photos"
os.makedirs(base_dir, exist_ok=True)

samples = [
    ("gauge1.png", "manometer_0 ~ 18.95"),
    ("gauge2.png", "manometer_1 ~ 42.10"),
    ("gauge3.png", "manometer_2 ~ 7.45"),
    ("gauge4.png", "manometer_3 ~ 63.30"),
    ("gauge5.png", "manometer_4 ~ 29.80"),
]

def make_img(path, text):
    w, h = 800, 600
    img = Image.new("RGB", (w, h), "white")
    d = ImageDraw.Draw(img)
    d.rectangle((20, 20, w-20, h-20), outline="black", width=4)
    d.ellipse((180, 100, w-180, h-100), outline="black", width=6)
    for i in range(0, 181, 15):
        x1 = 400 + 200 * math.cos(math.radians(180 - i))
        y1 = 300 + 200 * math.sin(math.radians(180 - i))
        x2 = 400 + 220 * math.cos(math.radians(180 - i))
        y2 = 300 + 220 * math.sin(math.radians(180 - i))
        d.line((x1, y1, x2, y2), fill="black", width=3)
    d.ellipse((390, 290, 410, 310), fill="black")
    d.line((400, 300, 570, 300), fill="red", width=5)
    d.text((40, 40), text, fill="black")
    img.save(path, format="PNG")

for fname, text in samples:
    make_img(os.path.join(base_dir, fname), text)

print("Created PNG files:")
for fname, _ in samples:
    print(os.path.join(base_dir, fname))
```
